```python
# 測試 GeoPandas 套件
try:
    import geopandas as gpd
    import pandas as pd
    import matplotlib.pyplot as plt
    import folium
    import osmnx as ox
    import numpy as np
    from shapely.geometry import Point
    
    print('✅ Conda GeoPandas 環境測試:')
    print(f'  geopandas {gpd.__version__}')
    print(f'  pandas {pd.__version__}')
    print(f'  matplotlib {matplotlib.__version__}')
    print(f'  folium {folium.__version__}')
    print(f'  osmnx {ox.__version__}')
    print(f'  shapely {gpd._compat.SHAPELY_VERSION}')
    print('\n🎉 所有套件載入成功！')
    
except ImportError as e:
    print(f'❌ 套件載入失敗: {e}')
    print('📦 嘗試安裝缺少的套件...')
```

```python
# 測試載入避難所資料
try:
    shelters_df = pd.read_csv('../data/shelters_cleaned.csv')
    shelters_gdf = gpd.GeoDataFrame(
        shelters_df,
        geometry=gpd.points_from_xy(shelters_df['經度'], shelters_df['緯度']),
        crs='EPSG:4326'
    )
    
    print(f'✅ 避難所資料載入成功:')
    print(f'  資料筆數: {len(shelters_gdf)}')
    print(f'  CRS: {shelters_gdf.crs}')
    print(f'  座標範圍: {shelters_gdf.total_bounds}')
    
    # 顯示前幾筆
    display(shelters_gdf.head(3))
    
except Exception as e:
    print(f'❌ 載入失敗: {e}')
```

```python
# 測試基本 GeoPandas 功能
# 創建台大土木系測試點
ntu_point = gpd.GeoDataFrame(
    {'name': ['台大土木系'], 'type': ['學校']},
    geometry=gpd.points_from_xy([121.5413], [25.0174]),
    crs='EPSG:4326'
)

print('📍 測試點創建:')
print(ntu_point[['name', 'type', 'geometry']])

# CRS 轉換測試
ntu_twd97 = ntu_point.to_crs(epsg=3826)
print(f'\n🔄 TWD97 轉換:')
print(f'  X: {ntu_twd97.geometry.x.iloc[0]:.2f} m')
print(f'  Y: {ntu_twd97.geometry.y.iloc[0]:.2f} m')

# 緩衝區測試
buffer_1km = ntu_twd97.buffer(1000)
print(f'\n🔵 1km 緩衝區面積: {buffer_1km.area.iloc[0]:.0f} 平方公尺')
print(f'   理論值: {3.14159 * 1000**2:.0f} 平方公尺')

print('\n✅ Conda GeoPandas 功能完全正常！')
```
