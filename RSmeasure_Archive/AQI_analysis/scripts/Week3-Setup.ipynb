### 1. 確認環境已安裝
我們已經在 Pre-lab 中安裝了所有必要的套件：

```python
# 檢查已安裝的套件版本
import geopandas as gpd
import pandas as pd
import matplotlib.pyplot as plt
import folium
import osmnx as ox
import numpy as np
from shapely.geometry import Point

print(f'geopandas {gpd.__version__} ✓')
print(f'pandas {pd.__version__} ✓')
print(f'folium {folium.__version__} ✓')
print(f'osmnx {ox.__version__} ✓')
print('所有套件已就緒！')
```

### 2. 載入避難所資料（使用 Week 2 的清理資料）

```python
# 載入 Week 2 清理後的避難所資料
shelters_df = pd.read_csv('../data/shelters_cleaned.csv')

# 轉換為 GeoDataFrame
shelters_gdf = gpd.GeoDataFrame(
    shelters_df,
    geometry=gpd.points_from_xy(shelters_df['經度'], shelters_df['緯度']),
    crs='EPSG:4326'
)

print(f'避難所資料載入成功: {len(shelters_gdf)} 筆')
print(f'CRS: {shelters_gdf.crs}')
print(f'座標範圍: {shelters_gdf.total_bounds}')
shelters_gdf.head()
```

### 3. 測試基本 GeoPandas 功能

```python
# 測試 CRS 轉換
shelters_twd97 = shelters_gdf.to_crs(epsg=3826)
print(f'轉換為 TWD97: {shelters_twd97.crs}')

# 測試緩衝區
buffer_1km = shelters_twd97.buffer(1000)  # 1公里緩衝區
print(f'1km緩衝區面積 (前5個):')
for i, area in enumerate(buffer_1km.head()):
    print(f'  避難所{i+1}: {area:.0f} 平方公尺')

print('\n✅ 所有 GeoPandas 功能正常運作')
```
